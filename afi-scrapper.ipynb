```python
# import libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd

response = requests.get("https://afi.com", headers={"User-Agent": "Mozilla/5.0"})
print(response.status_code)

# ran 200

```

```python
soup = BeautifulSoup(response.text, "html.parser")
print(soup.title.text)

# ran American Film Institute
```

```python
#display the first 500 characters of webpage code
print(response.text[:500])

# ran <!DOCTYPE html>...
```

```python
#look for standard link tags on page
links = soup.find_all("a")

#print the total number of links found
print(len(links))

# ran 183
```

```python
#loop through all 183 links found on the page
for link in links:
    #get the actual web address hidden inside the link
    href = link.get("href")

    #if the link exists and contains the word "picture" print
    if href and "picture" in href:
        print(href)

  # ran 3x https://www.afi.com/get-the-picture/
```

```python
#fetch the webpage data for game number 1000
test_response = requests.get("https://afi.com", headers={"User-Agent": "Mozilla/5.0"})
test_soup = BeautifulSoup(test_response.text, "html.parser")

#search the entire page layout for text or image assets
for tag in test_soup.find_all(["li", "img", "h3", "h4"]):
    text = tag.text.strip()
    #check if the element is the game image asset
    if tag.name == "img" and "gtp" in str(tag.get("class")):
        print(f"img:{tag.get('src')}")
    #check if the element holds the movie choice data
    elif "Guess the movie" in text or (tag.name == "li" and text.isupper()):
        print(f"{tag.name}:{text}")
```
