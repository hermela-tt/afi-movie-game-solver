```python

!pip install cloudscraper beautifulsoup4 pandas


import cloudscraper
from bs4 import BeautifulSoup
import pandas as pd
import time

URLS_TO_SCRAPE = ["https://www.afi.com/movie-games/508/"]

def scrape_afi_game_with_meta(url):
    scraper = cloudscraper.create_scraper(
        browser={
            'browser': 'chrome',
            'platform': 'windows',
            'desktop': True
        }
    )

    try:
        response = scraper.get(url)
        if response.status_code != 200:
            print(f"Skipping {url} (Status: {response.status_code})")
            return None

        soup = BeautifulSoup(response.text, 'html.parser')


        # Finding the game number from "Guess the movie (#508)
        game_title_element = soup.find('h2', class_='game-title') or soup.find('h1')
        game_title_text = game_title_element.text.strip() if game_title_element else ""

        # Finding the publish date/time if in   meta data
        date_element = soup.find('meta', property='article:published_time') or soup.find('time')
        # Checks if it can pull from  string, otherwise grabs the text
        game_date_time = date_element.get('content') or date_element.text.strip() if date_element else "Missing Date"

        # image pull line
        image_tag = soup.find('img', class_='game-image') or soup.find('div', class_='game-container').find('img')
        image_url = image_tag['src'] if image_tag else "Missing Image"

        incorrect_answers = []
        correct_answer = None
        user_percentages = {}

        options = soup.find_all('li', class_='game-option') or soup.find_all('div', class_='answer-option')

        for option in options:
            text_element = option.find('span', class_='option-text') or option
            answer_text = text_element.text.replace('\n', '').strip()

            pct_element = option.find('span', class_='vote-percentage') or option.find('span', class_='percent')
            pct_value = pct_element.text.strip() if pct_element else "0%"
            user_percentages[answer_text] = pct_value

            classes = option.get('class', [])
            is_correct = 'is-correct' in classes or 'correct' in classes or option.get('data-correct') == 'true'

            if is_correct:
                correct_answer = answer_text
            else:
                incorrect_answers.append(answer_text)

        # Packaging fields into the dictionary data row
        return {
            "game_header": game_title_text,
            "date_time_stamp": game_date_time,
            "source_url": url,
            "movie_image_url": image_url,
            "correct_answer": correct_answer,
            "incorrect_answers": incorrect_answers,
            "all_percentages": user_percentages
        }

    except Exception as e:
        print(f"Error handling URL {url}: {str(e)}")
        return None

# Loop execution
dataset = []
for target in URLS_TO_SCRAPE:
    data = scrape_afi_game_with_meta(target)
    if data:
        dataset.append(data)
    #wait 2 secs
    time.sleep(2)

df = pd.DataFrame(dataset)
df.head()

#error nonetype object
```

```python
import requests
import cloudscraper
from bs4 import BeautifulSoup
import pandas as pd
import time

start_day = 508
end_day = 508

def scrape_afi_game(day_number):
    base_url = "https://www.afi.com/movie-games/"
    url = base_url + str(day_number) + "/"

    # try cloudscraper first to slide past firewalls
    try:
        scraper = cloudscraper.create_scraper(
            browser={'browser': 'chrome', 'platform': 'windows', 'desktop': True}
        )
        response = scraper.get(url, timeout=10)
    except Exception:
        # fallback to standard requests if scraper crashes out completely
        try:
            headers = {
                'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
            }
            response = requests.get(url, headers=headers, timeout=10)
        except Exception as e:
            print(f"network link broke down totally {e}")
            return None

    if response.status_code != 200:
        print(f"site rejected the request status code is {response.status_code}")
        return None

    try:
        soup = BeautifulSoup(response.text, 'html.parser')

        # hunt down the game text title element
        header_tag = soup.find('h4')
        game_title = header_tag.text.strip() if header_tag else "guess game missing text"

        # pull the options from list blocks below heading
        choices = []
        if header_tag:
            options_list = header_tag.find_next('ul') or header_tag.find_next('ol')
            if options_list:
                choices = [li.text.strip() for li in options_list.find_all('li')]

        # track image source paths for the film shot
        image_url = "no image found"
        image_container = soup.find('div', class_='game-container') or soup.find('div', class_='wp-block-image')
        if image_container:
            img_tag = image_container.find('img')
            if img_tag and img_tag.get('src'):
                image_url = img_tag['src']

        # grab raw layout sources if container blocks miss it
        if image_url == "no image found":
            for img in soup.find_all('img'):
                src = img.get('src', '')
                if any(keyword in src for keyword in ['wp-content', 'movie-games', 'get-the-picture']):
                    image_url = src
                    break

        return {
            "game_number": day_number,
            "title_text": game_title,
            "image_link": image_url,
            "choices": choices,
            "url": url
        }
    except Exception as e:
        print(f"parsing error messed up layout processing {e}")
        return None

# run collection engine loops over the days
dataset = []
for day in range(start_day, end_day + 1):
    print(f"scraping day {day}")
    data = scrape_afi_game(day)
    if data:
        dataset.append(data)
    time.sleep(1)

# print dataset matrices out into colab layout
df = pd.DataFrame(dataset)
df.head()
```
